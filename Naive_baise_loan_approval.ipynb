```python
df['Gender']=df['Gender'].fillna(df['Gender'].mode()[0])
df['Married']=df['Married'].fillna(df['Married'].mode()[0])
df['Dependents']=df['Dependents'].fillna(df['Dependents'].mode()[0])
df['Self_Employed']=df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
df['Credit_History']=df['Credit_History'].fillna(df['Credit_History'].mode()[0])
df['Term']=df['Term'].fillna(df['Term'].mode()[0])
```

```python
df.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()

df['Gender'] = lab.fit_transform(df['Gender'])
df['Married'] = lab.fit_transform(df['Married'])
df['Dependents'] = lab.fit_transform(df['Dependents'])
df['Education'] = lab.fit_transform(df['Education'])
df['Self_Employed'] = lab.fit_transform(df['Self_Employed'])
df['Area'] = lab.fit_transform(df['Area'])
df['Status'] = lab.fit_transform(df['Status'])

# for column in df.columns:
#  if df[column].dtype == 'object':
#    df[column]=lab.fit_transform(df[column])
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

test train split

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

scaling

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

model creation

```python
from sklearn.naive_bayes import BernoulliNB
model = BernoulliNB()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
y_train
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,y_pred) * 100
score
```
